## Tipos de tensores


```python
import numpy as np
```

```python
# Tensor com apenas um número é chamado de escalar, ou tensor escalar, ou tensor de rank-0.

x_scalar = np.array(12)
print(x_scalar.ndim)
print(x_scalar.shape)
```

```python
# Vetores (rank-1)
# Array de números são tensores rank-1

x_vector = np.array([12, 3, 6, 14, 7]) # vetor de 5 dimensõ~es(entradas)
print(x_vector.ndim)
print(x_vector.shape)
```

```python
# Matrizes (tensores rank-2)

x_matrix = np.array([[5, 78, 2, 34, 0],
              [6, 79, 3, 35, 1],
              [7, 80, 4, 36, 2]],
)

print(x_matrix.ndim)
print(x_matrix.shape)
```

```python
# Tensores rank 3 e tensores de rank superiores
# Matrizes empacotadas, um cubo de números

x_matrix_3D = np.array(
    [[[5, 78, 2, 34, 0],
    [6, 79, 3, 35, 1],
    [7, 80, 4, 36, 2]],
    [[5, 78, 2, 34, 0],
     [6, 79, 3, 35, 1],
     [7, 80, 4, 36, 2]],
    [[5, 78, 2, 34, 0],
     [6, 79, 3, 35, 1],
     [7, 80, 4, 36, 2]]]

)
print(x_matrix_3D.ndim)
print(x_matrix_3D.shape)
```

## Atributos

- Eixos - rank - também podem ser chamados de tensores, informa a quantidade eixos que um tensor possui.
- Shape é uma tupla de dois inteiros, descreve quantos eixos e dimensões um tensor possui: `tensor.shape`.
- Dtype mostra qual é o tipo de dado que contém no tensor.
